```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('/Users/eric5805/Documents/1-YEAR-TRAINING/Projects/Car Sales/data/processed/processed.csv')
```

```python
#average price of cars by year
avg_price_by_year = df.groupby('Year of manufacture')['Price'].mean().round(2)

#average mileage of cars by year
avg_mileage_by_year = df.groupby('Year of manufacture')['Mileage'].mean().round(2)

#average millage of cars by brand
avg_mileage_by_brand = df.groupby('Manufacturer')['Mileage'].mean().round(2)

#average price of cars by brand
avg_price_by_brand = df.groupby('Year of manufacture')['Price'].mean().round(2)


```

```python
avg_year_info = pd.DataFrame([avg_price_by_brand])
avg_year_info
```

```python
plt.hist(df[df["Fuel type"]=="Petrol"]["Price"], bins=50, alpha=0.5, label="Petrol")
plt.hist(df[df["Fuel type"]=="Diesel"]["Price"], bins=50, alpha=0.5, label="Diesel")
plt.legend()
plt.title("Price Distribution by Fuel Type")
plt.xlabel("Price ($)")
plt.ylabel("Count")
plt.show()
```
