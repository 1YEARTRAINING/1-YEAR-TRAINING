2/Which area has the highest average house price overall?

```python
df.groupby('area')['average_price'].mean().sort_values().tail(1)
```

3/Which area has the lowest average house price overall?

```python
df.groupby('area')['average_price'].mean().sort_values().head(1)
```

4/ What is the mean, median, and standard deviation of house prices across all areas?

```python
df.average_price.describe().round(2)
```

5/ How many missing values exist in houses_sold and no_of_crimes?

```python
df[['houses_sold','no_of_crimes']].isna().sum()
```

6/ Convert date to datetime and extract year. What was the average house price trend by year across London?

```python
df.groupby('date')['average_price'].mean()
#df.date = df.date.dt.year
```

7/ Which year saw the biggest increase in average house prices?

```python
yearly_diff = df.groupby('date')['average_price'].mean().diff()
yearly_diff.idxmax()
```

8/ Compare the average number of houses sold per year — do you see a decline or growth trend?

```python
df.groupby('date').houses_sold.mean().round()
```

9/ For the borough "city of london", plot average house prices over time. Does it follow the same trend as the London-wide average?

```python
df.groupby('area').get_group('city of london')[["date", "average_price"]]
```

10/ Which borough had the highest number of crimes reported in any single year, and what was the corresponding average house price?

```python
df[['area','no_of_crimes','average_price']].sort_values(by = ['average_price','no_of_crimes'],ascending=[False, False]).head()
```
