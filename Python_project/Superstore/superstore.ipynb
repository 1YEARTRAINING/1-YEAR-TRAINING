```python
import pandas as pd
```

```python
# Detect encoding
import chardet

with open("superstore_ds.csv", "rb") as f:
    result = chardet.detect(f.read(50000))
```

```python
df = pd.read_csv("superstore_ds.csv", encoding=result["encoding"])
```

```python
df.head()
```

```python
df.columns
```

1/ How many unique customers are in the dataset?

```python
df['Customer ID'].nunique()
```

2/ Which category has the highest total sales?


```python
df.groupby('Category')['Sales'].sum().idxmax()
```

3/ What is the average discount applied across all orders?


```python
df.groupby('Order ID')['Discount'].mean()
```

4/ Convert Order Date to datetime. What year had the highest total profit?

```python
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Year'] = df['Order Date'].dt.year
```

```python
df.groupby('Year')['Profit'].sum().idxmax()
```

```python
df.groupby('Order Date')['Profit'].idxmax().sort_values(ascending=False).head()
```

5/ Which month tends to have the most orders?

```python
df['Month'] = df['Order Date'].dt.month
df.groupby('Month')['Order ID'].nunique().sort_values(ascending=False).head(1)
```

6/ Which customer has placed the most orders?

```python
df.groupby('Customer ID')['Order ID'].count().sort_values(ascending=False).head(1)
```

7/ Which segment contributes the highest profit margin (profit ÷ sales)?

```python
df.columns
```

```python
df['Segment'].value_counts()
```

```python
(df.groupby('Segment'))[['Sales','Profit']].sum().assign(Margin =lambda x : x['Profit']/x['Sales']).sort_values(by = 'Margin',ascending=False).head(1)
```

8/ Highest sales region

```python
df.columns
```

```python
df.groupby('Region')['Sales'].idxmax().sort_values(ascending=False)
```

9/Top 10 products by sales

```python
df.groupby('Product ID')[['Product Name','Sales']].sum().round(2).sort_values(by = 'Sales',ascending=False).head(10)
```

10/ Worst profit sub-category



```python
df.groupby('Sub-Category')[['Sub-Category','Profit']].sum(numeric_only= True).sort_values(by = 'Profit',ascending=True)
```

11/Average sales per order per state

```python
df.columns
```

```python
df.groupby('State')[['Sales','Order ID']].agg({'Sales' :'sum','Order ID' : 'nunique'}).assign(avg =lambda x: x['Sales']/x['Order ID']).sort_values(by = 'avg')
```

12/ Top 3 customers in each region by sales

```python
df['Customer ID'].nunique()
```

```python
(df.groupby(['Region','Customer ID'])['Sales'].sum()
                                            .round(2)
                                            .reset_index()
                                            .sort_values(by = ['Region','Sales'], ascending=[True,False])).groupby('Region').head(3)
```
